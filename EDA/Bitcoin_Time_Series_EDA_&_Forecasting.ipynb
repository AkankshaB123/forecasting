# Seasonality Decomposition
Separate Time Series Data into three components:


1.   **Trend** - General direction of time series
2.   **Seasonality** - Seasonal cycles (eg. summer, winter, cyclical)
* Additive: Seasonal cycles are constant in value
* Multiplicative: Seasonal cycles are proportional to the trend
* By understanding the seasonal cycles provides insights for time-series (How to identify: Interpret from the data visualisation itself & model performance)
3.   **Error Term** - What is not explained



```python
# Plotting the monthly seasonality
month_plot(df['Close'].resample('ME').mean(),
           ylabel = 'Closing')
plt.show()
# Red represent - average values; some sort of seasonality
```

```python
# Plotting the monthly seasonality
quarter_plot(df['Close'].resample('Q').mean(),
           ylabel = 'Closing')
plt.show()
# Red represent - average values, seasonality
```

# **Auto-correlation Plots**
* Whether there's information in the past to help us predict future
* The process is to correlate the time series with it's lagged values

```python
# Plot the auto-correlation (ACF)
fig, ax = plt.subplots(figsize=(12, 8))
plot_acf(df['Close'], lags=100, ax=ax)
plt.show()
```

Key points:
* Critical thinking: Forms actual linear curve - is it relevant?
* What is the unique information lag 100 brings that wasn't in lag 99 etc? (Sum up in Partial-autocorrelation)
* Most of it isn't relevant

```python
# Preview of choco_df; Plot the auto-correlation (ACF)
fig, ax = plt.subplots(figsize=(10, 6))
plot_acf(choco_sales_df['revenue'], lags=30, ax=ax)
plt.show()
# We need partial auto-correlation to accurately decipher this: PACF tells direct connection between data points at various times without the effect of inbetween points
# While ACF gives overall picture, PACF zooms in on specific
```

* Blue area is confidence interval (More in the past less relevant observation)

# **Partial Auto-Correlation**

Key points about PACF
1. Quick drop-off means only recent values directly impact current values.
2. Slow drop-off or oscillate means older values continue to exert a direct influence.
------------------------------

* Why PACF completes ACF: ACF begins by showing all correlations but may contain some indirect correlations too
* PACF concludes by highlighting direct correlations, offering a clearer understanding.

```python
# Plot the Partial Auto-Correlation (PACF)
fig, ax = plt.subplots(figsize=(10, 6))
plot_pacf(df['Close'], lags=100, ax=ax)
plt.show()
```

### Finding: Quick drop-off means only recent values directly impact current values

The only relevant information for the price of bitcoin is what happened the day before.

```python
# Preview of Choco df; Plot the Partial Auto-Correlation (PACF)
fig, ax = plt.subplots(figsize=(10, 6))
plot_pacf(choco_sales_df['revenue'], lags=30, ax=ax)
plt.show()
```

While ACF gives overall picture, PACF zooms in on specific

## LEARNINGS: What Went Wrong/Forecasts don't quite hit the mark
* The Rise & Fall of Fidget Spinners (JACKPOT for stores) - Mistakes a fleeting trend for a long lasting one
* Is this trend really here to stay or is it just a passing craze?
* Long-Term Capital Management's Mathematical Mishap, or LTCM: Hedge fund world - Russia Financial crisis (1998) - No matter how sophisticated models are there's stuff they can't predict
* The markets are a wild beast and sometimes they throw curveballs that no algorithm can catch
* Google Flu Trends: A Misstep in Big Data Predictions - Twist: Google Flu Trends ended up overestimating flu cases, sometimes way off the mark. Why? Well, it turns out that how & why people search for things can change a lot & Google's search algorithms kept changing too
* The Hidenburg Omen: An Ominous Market Predictor - airship disaster. Even the most complex & intriguing indicators can lead us astray if we rely too heavily on them without considering the bigger picture. Conclusion: Embracing the Unpredictability.
* No Model is 100% Correct
